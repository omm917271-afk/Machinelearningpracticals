## Step 1 : Divide dataframe into input and output Featutres
As we know Input X = {Pclass, Sex, Age, Fare, Embarked}
And Output Y = {Survived}

```python
import pandas as pd

data = pd.read_csv(r"C:\Users\omm91\Downloads\drug200.csv")
```

```python
# dataset.iloc [row_index,column_index]
# dataset.iloc[: , :-1]

X = data.iloc[:, :-1]
Y = data.iloc[:, -1]
```

```python
X.head()
```

```python
Y
```

### Step 2: Handle the missing values in Dataset

**From Above EDA we know two colomns `AGE` and `Embarked` contains the Missing Value**

1. Age -  Contineous - float
2. Embarked -  Categorical - Object - 3 Class - Most Frequent is Class {s}

```python
X.isnull().sum()
```

```python
from sklearn.impute import SimpleImputer

imputer = SimpleImputer(missing_values=np.nan,
                        strategy='most_frequent')

X.iloc[:, 2] = imputer.fit_transform(X.iloc[:, [2]]).ravel()
```

## Check Missing Values Solved in Age 

```python
X.isnull().sum()
```

```python
from sklearn.impute import SimpleImputer
imputer = SimpleImputer(missing_values=np.nan, 
                        strategy="most_frequent")
X.iloc[:, [-1]] = imputer.fit_transform(X.iloc[:, [-1]])
```

```python
X.isnull().sum()
```

## Step 3 : Encoding categorical data

**Types of Encoding**
1.  Label Encoder -  USed when feature contains 2 categories
2. One Hot Encoder -  Used when more than two categories

```python
X['Sex'].value_counts()
```

**As it contains two categories apply label encoder**

```python
X['Na_to_K'].value_counts()
```

**It contains three categories hence used One Hot Encoder**

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
X['Sex'] = le.fit_transform(X['Sex'])
```

```python
X.head()
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

ct = ColumnTransformer(transformers=[
    ('encoder', OneHotEncoder(), [0, 1])
], remainder='passthrough')
```

```python
X = np.array(ct.fit_transform(X))
print(X)
```

### Step 4 : Splitting Data into Training and Testing

```python
from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(
    X,
    Y,
    test_size=0.3,
    random_state=0
)
```

```python
print(X_train.shape)
```

### Step 5 : Feature Scaling

```python
from sklearn.preprocessing import StandardScaler

sc = StandardScaler()

X_train[:, 5:] = sc.fit_transform(X_train[:, 5:])
X_test[:, 5:] = sc.transform(X_test[:, 5:])
```
