## Import Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

## Preprocessing Steps

```python
# Step 1 : Seprate i/p Independent Var and 
# o/p Dependent Var
X = dataset[["math score"]]
y = dataset["writing score"]
```

```python
print(X.head())
```

```python
print(y.head())
```

```python
# Step 4 : Split data into training and testin
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test = train_test_split (X,y,
                                        test_size = 0.2,
                                        random_state = 30)
```

```python
print(X_train.shape)
print(X_test.shape)
```

## Create the Regression model on training data

```python
from sklearn.linear_model import LinearRegression

regressor = LinearRegression()

regressor.fit(X_train,y_train)
```

```python
print("Intercept B0 = ",regressor.intercept_)
print("Coefficient B1= ",regressor.coef_)
```

From Above values our model regression line equation

**y = 25566.43561640947 + 9481.03756369 * X**

By using this equation lets find salary for Experince =6

```python
math_score = 80
ynew = regressor.intercept_ + regressor.coef_[0] * math_score

print("Predicted writing score for math score = 80: ", ynew)
```

```python
a = pd.DataFrame([[80.0]], columns=["math score"])
ynew = regressor.predict(a)
print(ynew)
```

```python
ypred = regressor.predict(X_test)
```

```python
print(ypred) #predicted value vector\

print(y_test) # actual values
```

#### Accuracy of model

```python
# 1. Training Accuracy

print("Training accuracy =",
      regressor.score(X_train,y_train))
```

```python
# 2. Testing Accuracy

print("Testing accuracy =",
      regressor.score(X_test,y_test))
```

### Visualizing the Model

#### 1. Visualize the training data

```python
plt.scatter(X_train, y_train, color = 'red', s = 12)
x_line = np.linspace(X_train["math score"].min(), X_train["math score"].max(), 100)
x_line_df = pd.DataFrame(x_line, columns=["math score"])
plt.plot(x_line, regressor.predict(x_line_df), color = 'blue')
plt.title("Simple Regression on Training Data", size = 16)
plt.xlabel("Math Score")
plt.ylabel("Writing Score")
plt.show()
```

#### 2. Visualize the testing data

```python
plt.scatter(X_test, y_test, color = 'red', s = 12)
x_line = np.linspace(X_test["math score"].min(), X_test["math score"].max(), 100)
x_line_df = pd.DataFrame(x_line, columns=["math score"])
plt.plot(x_line, regressor.predict(x_line_df), color = 'blue')
plt.title("Simple Regression on Testing Data", size = 16)
plt.xlabel("Math Score")
plt.ylabel("Writing Score")
plt.show()
```

#### Find R^2 Score of model

```python
from sklearn.metrics import r2_score
r2_score(y_test,ypred)
```
